```python
 #MBC 마켓에서 이번엔 럭키백을 패션분야에 접목을 시키려고 한다.

# 패션 분야이기 때문에 상품 개수가 늘어나고 어종이 아닌
# 일반 잡화용으로 제공하기로 하였다.

# 딥러닝의 시작 인공 신경망

# MNIST 데이터 셋 : 손으로 쓴 0~9까지의 숫자로 이루어져 있음
# 패션 MNIST 데이터 셋 : 숫자대신 패션아이템으로 이루어짐
# https://www.kaggle.com/datasets 캐글에가면 연습용 데이터 셋이 많다.
# https://www.aihub.or.kr/ 국내에서 제공하는 연습용 데이터 셋 & 동영상강의
```

```python
import tensorflow as tf
from tensorflow import keras

# 랜덤 고정
tf.keras.utils.set_random_seed(42)

# Fashion MNIST 데이터셋 불러오기
(train_input, train_target), (test_input, test_target) = keras.datasets.fashion_mnist.load_data()

print("===== Raw Data Shape =====")
print("Train Input:", train_input.shape)
print("Train Target:", train_target.shape)
print("Test Input:", test_input.shape)
print("Test Target:", test_target.shape)

# 이미지 픽셀값 스케일링 (0~255 ➜ 0~1 범위)
train_input = train_input / 255.0
test_input = test_input / 255.0

print("\n===== Scaled Data Sample =====")
print(train_input[0])

# target 데이터 확인
print("\n===== Unique Target Labels =====")
print(set(train_target))
```

```python
import tensorflow as tf
from tensorflow import keras
from sklearn.model_selection import train_test_split

# 데이터 로드
(train_input, train_target), (test_input, test_target) = keras.datasets.fashion_mnist.load_data()

# 데이터 스케일링 (정규화)
train_input = train_input / 255.0
test_input = test_input / 255.0

# 훈련세트(80%) - 검증세트(20%) 분리
train_input, val_input, train_target, val_target = train_test_split(
    train_input, train_target, test_size=0.2, random_state=42
)

# Shape 출력
print("===== Dataset Shapes =====")
print("Train input shape:", train_input.shape)
print("Train target shape:", train_target.shape)
print("Validation input shape:", val_input.shape)
print("Validation target shape:", val_target.shape)
print("Test input shape:", test_input.shape)
print("Test target shape:", test_target.shape)
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# 모델 생성 (Sequential)
model = keras.Sequential([
    layers.Conv2D(32, (3,3), activation='relu', input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2,2)),
    layers.Conv2D(64, (3,3), activation='relu'),
    layers.MaxPooling2D((2,2)),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')   # Output layer (10 classes)
])

# 모델 구성 확인 (화면 캡쳐 포인트!)
model.summary()
```

```python
# 다중 분류 모델 컴파일
model.compile(
    optimizer='adam',                         # 옵티마이저
    loss='sparse_categorical_crossentropy',    # 손실 함수 (정수 라벨용)
    metrics=['accuracy']                       # 평가 지표
)

print("===== Model Compile Completed ====")
```

```python
# 모델 학습 (fit)
history = model.fit(
    train_input.reshape(-1, 28, 28, 1),  # CNN 입력 형태로 reshape
    train_target,
    epochs=10,                           # 학습 반복 횟수
    batch_size=32,                       # 배치 크기
    validation_data=(val_input.reshape(-1, 28, 28, 1), val_target)  # 검증 데이터 포함
)

print("===== Training Completed ====")
```

```python
#최적화 기법
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint

# EarlyStopping: 더 이상 좋아지지 않을 때 학습 중단
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=3,
    restore_best_weights=True
)

# ReduceLROnPlateau: 학습률 자동 감소
lr_scheduler = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=2
)

# ModelCheckpoint: 최고 성능 모델 자동 저장
checkpoint = ModelCheckpoint(
    filepath='best_model.keras',
    monitor='val_accuracy',
    save_best_only=True
)

# 모델 학습 (최적화 기법 적용)
history = model.fit(
    train_input.reshape(-1, 28, 28, 1),
    train_target,
    epochs=20,
    batch_size=32,
    validation_data=(val_input.reshape(-1, 28, 28, 1), val_target),
    callbacks=[early_stop, lr_scheduler, checkpoint]
)

print("===== Optimization Training Completed ====")
```

```python
#과대적합
from tensorflow import keras
from tensorflow.keras import layers, regularizers

model = keras.Sequential([
    layers.Conv2D(32, (3,3), activation='relu', input_shape=(28,28,1)),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2,2)),

    layers.Conv2D(64, (3,3), activation='relu', kernel_regularizer=regularizers.l2(0.001)),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2,2)),

    layers.Flatten(),
    layers.Dropout(0.5),         # *** 과대적합 방지 핵심 ***
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

model.summary()

print("===== Overfitting Prevention Model Created =====")
```

```python
# 모델 요약 출력
model.summary()

print("===== Model Summary Completed =====")
```

```python
import matplotlib.pyplot as plt

# 학습 이력에서 값 가져오기
acc = history.history['accuracy']
val_acc = history.history['val_accuracy']
loss = history.history['loss']
val_loss = history.history['val_loss']

epochs = range(1, len(acc) + 1)

# 정확도 그래프
plt.figure(figsize=(8,4))
plt.plot(epochs, acc, label='Train Accuracy')
plt.plot(epochs, val_acc, label='Validation Accuracy')
plt.title('Training and Validation Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)
plt.show()

# 손실 그래프
plt.figure(figsize=(8,4))
plt.plot(epochs, loss, label='Train Loss')
plt.plot(epochs, val_loss, label='Validation Loss')
plt.title('Training and Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)
plt.show()

print("===== Training Graphs Displayed =====")
```

```python
# 모델 저장
model.save("fashion_cnn_model.h5")

print("===== Model Saved Successfully =====")
```

```python
import tensorflow as tf
from tensorflow import keras

# 저장된 모델 불러오기
loaded_model = keras.models.load_model('fashion_cnn_model.h5')

# 모델 컴파일 (평가하기 위해 필요)
loaded_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# 테스트 데이터로 모델 평가
# test_input과 test_target은 이전에 로드되고 스케일링된 데이터를 사용합니다.
# CNN 입력 형태에 맞게 reshape합니다.
test_loss, test_accuracy = loaded_model.evaluate(test_input.reshape(-1, 28, 28, 1), test_target, verbose=0)

print(f'Test Loss: {test_loss:.4f}')
print(f'Test Accuracy: {test_accuracy*100:.2f}%')
```
